# No Poker Face — analysis notebook
Loads one run folder and reproduces the report tables. Set `RUN` to `calibration`, `main`, or a `_mock` folder (mock folders hold synthetic outputs, never data).

```
import sys, json; sys.path.insert(0, '..')
from pathlib import Path
import pandas as pd
from npf.analysis import summarize, simulate_power
RUN = Path('../runs/calibration_mock')
df, sl, res = summarize(RUN)
print('items', len(df))
```

```
pd.DataFrame(res['by_condition']).query("domain=='all'")
```

```
pd.DataFrame({k: v['all'] for k, v in res['contrasts'].items()}).T
```

```
import matplotlib.pyplot as plt
t = sl[sl['mode']=='options_only'].groupby(['domain','cond']).correct.mean().unstack(0)
t.plot.bar(figsize=(8,3.5), ylabel='options-only accuracy', title='Options-only accuracy by condition'); plt.axhline(.25, ls=':', c='grey'); plt.tight_layout()
```

```
p = simulate_power(sl)
pd.DataFrame(p['grid']).pivot_table(index=['slots','samples_per_cell'], columns='effect_pts', values='power')
```
